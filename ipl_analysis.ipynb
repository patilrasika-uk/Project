```python
import matplotlib.pyplot as plt
import pandas as pd
df = pd.read_csv(
    "ipl_dummy_dataset.csv",

)

print(df.to_string())
print(df.info())
```

```python
top5=df.groupby('winner')['wickets'].count().head(5)
top5.plot(kind="line", marker='8',ms=20)
plt.title('total wickets by cities')
plt.xlabel('Winners')
plt.ylabel('wickets')
plt.xticks(rotation=45)
plt.show()
```

```python
df['venue'].value_counts().head(5).plot.pie(autopct='%1.1f%%')
plt.title("top five city")
plt.show()
```

```python
plt.figure(figsize=(10,5))
wickets_count=df.groupby("winner")["runs"].sum().sort_values(ascending=False)
wickets_count.plot(kind='bar')
plt.title('Top wikets by runs')
plt.xlabel('count')
plt.ylabel('Total Runs')
plt.show()
```

```python
toss_decision_count = df['toss_decision'].value_counts()
plt.figure(figsize=(5,4))
toss_decision_count.head(10).plot(kind='bar',color="gray")
plt.title('toss_winner')
plt.xlabel('toss')
plt.ylabel('winner')
```

```python
df['winner'].value_counts().plot(kind="pie")
plt.title('toss_winner')
plt.xlabel('toss')
plt.ylabel('winner')
plt.show()
```

```python
newdf=df.groupby('winner')['toss_winner'].count().sort_values(ascending=False)
newdf.plot(kind='bar',color='g')
plt.title('winner vs toss')
plt.xlabel('toss winner')
plt.ylabel('winner')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Create figure
plt.figure(figsize=(18, 15))

# ---------------------------------------------------
# 1. Total wickets by winners (Line Plot)
# ---------------------------------------------------
plt.subplot(2, 3, 1)

top5 = df.groupby('winner')['wickets'].count().head(5)

top5.plot(
    kind='line',
    marker='o',
    markersize=10,
    linewidth=2,
    color='blue'
)

plt.title('Total Wickets by Winners', fontsize=14)
plt.xlabel('Winners')
plt.ylabel('Wickets')
plt.xticks(rotation=45)

# ---------------------------------------------------
# 2. Top 5 Venues (Pie Chart)
# ---------------------------------------------------
plt.subplot(2, 3, 2)

df['venue'].value_counts().head(5).plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Top 5 Venues', fontsize=14)
plt.ylabel('')

# ---------------------------------------------------
# 3. Total Runs by Winners (Bar Chart)
# ---------------------------------------------------
plt.subplot(2, 3, 3)

runs_count = df.groupby('winner')['runs'].sum().sort_values(ascending=False)

runs_count.plot(
    kind='bar',
    color='orange'
)

plt.title('Total Runs by Winners', fontsize=14)
plt.xlabel('Winner')
plt.ylabel('Total Runs')
plt.xticks(rotation=45)

# ---------------------------------------------------
# 4. Toss Decision Count (Bar Chart)
# ---------------------------------------------------
plt.subplot(2, 3, 4)

toss_decision_count = df['toss_decision'].value_counts()

toss_decision_count.plot(
    kind='bar',
    color='gray'
)

plt.title('Toss Decision Count', fontsize=14)
plt.xlabel('Toss Decision')
plt.ylabel('Count')
plt.xticks(rotation=0)

# ---------------------------------------------------
# 5. Winner Distribution (Pie Chart)
# ---------------------------------------------------
plt.subplot(2, 3, 5)

df['winner'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Winner Distribution', fontsize=14)
plt.ylabel('')

# ---------------------------------------------------
# 6. Winner vs Toss Winner (Bar Chart)
# ---------------------------------------------------
plt.subplot(2, 3, 6)

winner_toss = df.groupby('winner')['toss_winner'].count().sort_values(ascending=False)

winner_toss.plot(
    kind='bar',
    color='green'
)

plt.title('Winner vs Toss Winner', fontsize=14)
plt.xlabel('Winner')
plt.ylabel('Toss Wins')
plt.xticks(rotation=45)

# Adjust layout
plt.tight_layout()

# Show plots
plt.show()
```
