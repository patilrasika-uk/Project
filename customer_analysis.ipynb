```python
df['Email Domain']=df ["Email"].str.split ('@').str[1]
print (df['Email Domain'])
```

```python
#10. Website Analysis
#Count .com, .org, .net websites.
#Find most common website domain.
print ("com:", df["Website"].str.contains(".com", na=False).sum())
print ("org:", df["Website"].str.contains(".org", na=False).sum())
print ("net websites:", df["Website"].str.contains("net websites", na=False).sum())
```

```python
#printart C — GroupBy Practice
#11. Country-wise Customer Count
country_count= df.groupby("Country")["Customer Id"].count()
print(country_count)
```

```python
#12. City-wise Analysis
#Find:
#Top 10 cities by customer count.
city_count = df .groupby("City")["Customer Id"].count()
print(city_count)
```

```python
#13. Company Analysis
#Find:
#Company with most customers.
#Top 5 companies.
top_5_companies = df['Company'].value_counts().head(5)
print(top_5_companies)
```

```python
df['Subscription Date'] = pd.to_datetime(df['Subscription Date'],dayfirst=True)
print(df['Subscription Date'])
```

```python
#14. Subscription Analysis
#Find:
#Customers joined each year.
df['Year'] = df['Subscription Date'].dt.year
print(df['Year'].value_counts().sort_index())
```

```python
#Customers joined each month.
df['Month'] = df['Subscription Date'].dt.month_name()
print(df['Month'].value_counts().sort_index())
```

```python
#15. Email Provider Analysis
#Create chart for:
#Gmail users
#Yahoo users
#Outlook users
import matplotlib.pyplot as plt 
Provider_count = {
    'gmail':(df['Email Domain'] == "gmail.com").sum(),
    'yahoo':(df['Email Domain'] == "yahoo.com").sum(),
    'outlook':(df['Email Domain'] == "outlook.com").sum()
}

plt.figure(figsize=(6,4))
plt.bar(Provider_count.keys(),Provider_count.values())
plt.title('Email Provider Analysis')
plt.xlabel('provider')
plt.ylabel('Users')
plt.show()


```

```python
# Part D — Matplotlib Practice
#16. Bar Chart
#Plot:
#Top 10 cities by customers

plt.figure(figsize=(10,5))
city_count.head(10).plot(kind='bar') 
plt.title('Top 10 cities by customers')
plt.xlabel('City')
plt.ylabel('Customer')
plt.show()
```

```python
#17. Pie Chart
# Show:
# Country distribution.

plt.figure(figsize=(8,5))
country_count.plot(kind='pie', autopct='%1.1f%%')
plt.title('Country Distribution')
plt.ylabel('')
plt.show()
```

```python
#18. Line Chart
# Plot:
# Number of subscriptions over time

subcription_trend = df.groupby('Subscription Date')['Index'].count()

plt.figure(figsize=(10,5))
plt.plot(subcription_trend)
plt.title('subscriptions Over Time')
plt.xlabel('Date')
plt.ylabel('Customers')
plt.show()
```

```python
df['Month'] = df['Subscription Date'].dt.month_name()

# Count subscriptions every month
monthly_subscriptions = df['Month'].value_counts()

print(monthly_subscriptions)
```

```python
month_order = [
    'January', 'February', 'March', 'April', 
]

monthly_subscriptions = df['Month'].value_counts().reindex(month_order)

print(monthly_subscriptions)
```

```python

plt.figure(figsize=(10,5))

monthly_subscriptions.plot(kind='bar')

plt.title('Number of Subscriptions Per Month')
plt.xlabel('Month')
plt.ylabel('Subscriptions')

plt.show()
```

```python
#20. Horizontal Bar Chart
# Plot:
# Top companies by customer count

company_count = df['Company'].value_counts()
plt.figure(figsize=(10,6))
company_count.head(10).plot(kind='barh')
plt.title('Top companies by customer count')
plt.xlabel('Customers')
plt.ylabel('Company')
```

```python
#🔹 Part E — Real Data Science Task
# 21. Customer Growth Analysis
# Find customer growth trend over months.
monthly_growth = df.groupby(['year','month'])['Index'].count()
print("\nCUSTOMERS GROWTH TREND")
print(monthly_growth)
```

```python
#22. Business Insight Questions
# AnswerS
# Which country has most customers?
# Which company has highest registrations?
# Which email provider is most popular?
# Country with most customers
print("\nCOUNTRY WITH MOST CUSTOMERS")
print(country_count.idxmax())
# Company with highest registrations 
print("\nCOMPANY WITH HIGHEST REGISTRATIONS") 
print(company_count.idxmax()) 
# Most popular email provider 
print("\nMOST POPULAR EMAIL PROVIDER")
print(df['Email Domain'].mode())
```

```python
#23. Detect Invalid Data
#Find:
#Invalid emails
#Missing websites
#Incorrect phone numbers
# Invalid emails 
invalid_emails = df[~df['Email'].str.contains('@', na=False)] 
print("\nINVALID EMAILS") 
print(invalid_emails)

# Missing websites 
missing_websites = df[df['Website'].isnull()]
print("\nMISSING WEBSITES") 
print(missing_websites) 

# Incorrect phone numbers 
df['Phone 1'] = df['Phone 1'].astype(str)
incorrect_phone = df[df['Phone 1'].str.len() < 10]
print("\nINCORRECT PHONE NUMBERS") 
print(incorrect_phone)
```

```python
#24. Customer Segmentation
#Classify customers based on:
#Country
#Company
#Email domain
# Segment by Country 
country_segment = df.groupby('Country')['Index'].count() 

# Segment by Company 
company_segment = df.groupby('Company')['Index'].count() 

# Segment by Email Domain
email_segment = df.groupby('Email Domain')['Index'].count()
print("\nCOUNTRY SEGMENT") 
print(country_segment) 
print("\nCOMPANY SEGMENT") 
print(company_segment) 
print("\nEMAIL SEGMENT") 
print(email_segment)
```
