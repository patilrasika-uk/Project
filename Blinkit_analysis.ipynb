```python
import pandas as pd 
import matplotlib.pyplot as plt 

df=pd.read_csv('blinkit_store_data (1).csv')
print (df.info())
print (df.shape)
```

```python
#1. Find the top 10 cities by total revenue.
print(df.groupby('City')['Revenue'].sum().sort_values(ascending=False).head(10))
```

```python
#. Which product category generated the highest sales?
print(df.groupby('Category')['Revenue'].sum().max())
```

```python
#Calculate average delivery time city-w'ise.
avrg_delivery = df.groupby('City')['Delivery_Time_Minutes'].mean()
print(avrg_delivery)
```

```python
#. Find the total number of orders for each payment method.
total_orders =df.groupby('payment').value_counts.sort_values(ascending=False) 
total_orders =df.groupby('payment').value_counts.sort_values(ascending=False) 
print()
```
